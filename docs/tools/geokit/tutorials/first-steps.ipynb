# First steps with geokit

In this tutorial you compute distances between cities, find a bearing and measure a polygon.

This page is a Jupyter notebook. It is committed **with its outputs**, and the docs build does not re-run it.

## 1. Distance between two cities

```python
from geokit import haversine, bearing

paris = (48.8566, 2.3522)
london = (51.5074, -0.1278)

print(f"Paris to London: {haversine(*paris, *london):.1f} km")
```

## 2. Which way to go

```python
print(f"Initial bearing: {bearing(*paris, *london):.1f} degrees")
```

## 3. Distances from Paris to several cities

```python
import matplotlib.pyplot as plt

cities = {
    "London": (51.5074, -0.1278),
    "Brussels": (50.8503, 4.3517),
    "Berlin": (52.52, 13.405),
    "Madrid": (40.4168, -3.7038),
    "Rome": (41.9028, 12.4964),
}
km = {name: haversine(*paris, *pos) for name, pos in cities.items()}

fig, ax = plt.subplots(figsize=(6, 3))
ax.barh(list(km), list(km.values()))
ax.set_xlabel("km from Paris")
fig.tight_layout()
```

## 4. Area of a polygon

```python
from geokit import Point, Polygon

square = Polygon([Point(0, 0), Point(2, 0), Point(2, 2), Point(0, 2)])
square.area(), square.perimeter()
```
